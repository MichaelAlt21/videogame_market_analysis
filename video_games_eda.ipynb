### Описание данных
Данные `/datasets/new_games.csv` содержат информацию о продажах игр разных жанров и платформ, а также пользовательские и экспертные оценки игр:

- `Name` — название игры.
- `Platform` — название платформы.
- `Year of Release` — год выпуска игры.
- `Genre` — жанр игры.
- `NA sales` — продажи в Северной Америке (в миллионах проданных копий).
- `EU sales` — продажи в Европе (в миллионах проданных копий).
- `JP sales` — продажи в Японии (в миллионах проданных копий).
- `Other sales` — продажи в других странах (в миллионах проданных копий).
- `Critic Score` — оценка критиков (от 0 до 100).
- `User Score` — оценка пользователей (от 0 до 10).
- `Rating` — рейтинг организации ESRB (англ. Entertainment Software Rating Board). Эта ассоциация определяет рейтинг компьютерных игр и присваивает им подходящую возрастную категорию.

- **RP (Rating Pending)** — используется в рекламных материалах для игр, которым ещё не присвоен окончательный рейтинг ESRB. Принят в 1994 году и действует до сих пор.
- **EC (Early Childhood)**— игры, ориентированные на дошкольную аудиторию. Не содержат материалов, которые родители могли бы считать неприемлемыми. Принят в 1994 году и действовал до 2018 года.
- **E (Everyone)** — игры для всех возрастов. Могут содержать незначительное количество насилия и сленга. До 1998 года рейтинг был известен как Kids to Adults (K-A). Принят в 1994 году и действует до сих пор.
- **E10+** **(Everyone 10 and older)** — игры для людей от 10 лет и старше. Могут содержать более высокий уровень насилия, грубую разговорную речь. Принят в 2005 году и действует до сих пор.
- **T (Teen)** — игры для людей от 13 лет и старше. Могут содержать умеренное количество насилия, включая небольшие количества крови, грубую разговорную речь и юмор, сексуальное содержание. Принят в 1994 году и действует до сих пор.
- **M (Mature)** — игры для лиц от 17 лет и старше. Могут содержать более интенсивные или реалистичные изображения насилия, включая кровь, ужасы, увечья и изображения смерти, непристойные сексуальные сцены, частичную наготу и грубую ненормативную лексику. Принят в 1994 году и действует до сих пор.
- **AO (Adults Only)**— игры для людей от 18 лет и старше. Содержат материалы с более высоким уровнем воздействия, чем рейтинг M. Принят в 1994 году и действует до сих пор.

### 2.3. Наличие пропусков в данных

```python
print(df.isnull().sum())
print(sum(df.isnull().sum()))
```

```python
print(df.isnull().mean())
```

```python
def show_missing_stats(tmp0):
    """
    Функция для отображения статистики пропущенных значений в DataFrame.
    """
    missing_stats = pd.DataFrame({
        'Кол-во пропусков': tmp0.isnull().sum(),
        'Доля пропусков': tmp0.isnull().mean()
    })
    missing_stats = missing_stats[missing_stats['Кол-во пропусков'] > 0]
    
    if missing_stats.empty:
        return "Пропусков в данных нет"
    
    return (missing_stats.style.format({'Доля пропусков': '{:.4f}'}).background_gradient(cmap='coolwarm'))
show_missing_stats(df)
```

```python
print(df.isnull().mean() * 100)
```

40-54% пропусков у оценок с рейтингами - вероятно проблемы со сбором данных, либо слишком много различных площадок, где у игры могут быть абсолютно разные оценки. В остальных столбцах максимум 1.6% пропусков. Строки с пропущенными именем, жанром, годом релиза стоит удалить - это важные параметры, которые определяют игру. С пропущенной информацией о продажах заменить на среднее значение в зависимости от названия платформы, жанра и года выхода игры, ведь можно представить насколько была популярна такая игра в определенное время. Пропущенные оценки и рейтинг скорее всего заменить на индикатор. 

```python
df_clean = df.dropna(subset=['name', 'genre', 'year_of_release']).copy()

print(f"Удалено строк с пропущенными name/genre/year: {len(df) - len(df_clean)}")
print(f"Осталось строк: {len(df_clean)}")
```

```python
sales_cols = ['eu_sales', 'jp_sales']
for col in sales_cols:
    avg_sales = df_clean.groupby(['platform', 'year_of_release'])[col].transform('mean')
    df_clean[col] = df_clean[col].fillna(avg_sales)
```

```python
df_clean['critic_score'] = df_clean['critic_score'].fillna(-1)
df_clean['user_score'] = df_clean['user_score'].fillna(-1)
df_clean['rating'] = df_clean['rating'].fillna('unknown')
```

```python
print(df_clean.isnull().sum())
print(f"Осталось строк: {len(df_clean)}")
```

```python
key_cols = ['name', 'platform', 'year_of_release']
duplicate_groups = df_clean[df_clean.duplicated(subset=key_cols, keep=False)]

if len(duplicate_groups) > 0:
    print(f"\nНайдены группы записей с одинаковыми {key_cols}:")
    for _, group in duplicate_groups.groupby(key_cols):
        if len(group) > 1:
            print(f"\n  Группа: {group.iloc[0]['name']} - {group.iloc[0]['platform']} - {group.iloc[0]['year_of_release']}")
            print(f"    Количество записей: {len(group)}")
            print(f"    Продажи (NA, EU, JP, Other):")
            for _, row in group.iterrows():
                print(f"      {row['na_sales']}, {row['eu_sales']}, {row['jp_sales']}, {row['other_sales']}")
    
    print("\nОставляем запись с максимальными общими продажами для каждой группы")
    
    df_clean['total_sales'] = df_clean['na_sales'] + df_clean['eu_sales'] + df_clean['jp_sales'] + df_clean['other_sales']
    df_clean = df_clean.sort_values('total_sales', ascending=False).drop_duplicates(subset=key_cols, keep='first')
    df_clean = df_clean.drop('total_sales', axis=1)
    
    print(f"После обработки осталось строк: {len(df_clean)}")
else:
    print("\nАномальных дубликатов не найдено")
```

## 4. Категоризация данных

```python
def categorize_user_score(score):
    if score == -1:
        return 'неизвестно'
    elif 8 <= score <= 10:
        return 'высокая оценка'
    elif 3 <= score < 8:
        return 'средняя оценка'
    elif 0 <= score < 3:
        return 'низкая оценка'
    else:
        return 'неизвестно'

df_actual['user_score_category'] = df_actual['user_score'].apply(categorize_user_score)

print(df_actual['user_score_category'].value_counts())
```

```python
def categorize_critic_score(score):
    if score == -1:
        return 'неизвестно'
    elif 80 <= score <= 100:
        return 'высокая оценка'
    elif 30 <= score < 80:
        return 'средняя оценка'
    elif 0 <= score < 30:
        return 'низкая оценка'
    else:
        return 'неизвестно'

df_actual['critic_score_category'] = df_actual['critic_score'].apply(categorize_critic_score)

print(df_actual['critic_score_category'].value_counts())
```

```python
top_platforms = df_actual['platform'].value_counts().head(7)

print("Топ-7 платформ по количеству игр:")
print(top_platforms)
print("\nВ процентном соотношении:")
print((top_platforms / len(df_actual) * 100).round(2))
```
